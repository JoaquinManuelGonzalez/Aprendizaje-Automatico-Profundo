```python
import matplotlib.pyplot as plt

imagen = plt.imread('hoja.jpg')
plt.figure()
plt.imshow(imagen) 
```

```python
import matplotlib.pyplot as plt

# Visualizar la imagen original
plt.figure(figsize=(4,4))
plt.imshow(imagen)
plt.title('Imagen Original')
plt.axis('off')
plt.show()
```

```python
C = ['Reds', 'Greens', 'Blues']
for p in [0,1,2]:
    plt.figure(figsize=(4,4)) 
    plt.imshow (imagen[:,:,p],cmap=C[p]) 
    plt.show()
```
